```python
import pandas as pd
```

```python
air_1_df = pd.read_csv("../data/2016_1.csv", encoding='EUC-KR')
air_2_df = pd.read_csv("../data/2016_2.csv", encoding='EUC-KR')
air_3_df = pd.read_csv("../data/2016_3.csv", encoding='EUC-KR')
air_4_df = pd.read_csv("../data/2016_4.csv", encoding='EUC-KR')
```

```python
air_df = pd.concat([air_1_df, air_2_df, air_3_df, air_4_df])
```

```python
air_df.columns
```

```python
from collections import Counter
Counter(air_df['지역'])
```

```python
interested_places = {'BS': '부산',
         'CB': '충북',
         'CN': '충남',
         'DG': '대구',
         'DJ': '대전',
         'GB': '경북',
         'GG': '경기',
         'GJ': '광주',
         'GN': '경남',
         'GW': '강원',
         'IC': '인천',
         'JB': '전북',
         'JJ': '제주',
         'JN': '전남',
         'SU': '서울',
         'US': '울산'}
```

```python
import datetime
def my_to_datetime(date_str):
    date_str = str(date_str)
    if date_str[8:10] != '24':
        return pd.to_datetime(date_str, format='%Y%m%d%H').date()

    date_str = date_str[0:8] + '00' + date_str[10:]
    return pd.to_datetime(date_str, format='%Y%m%d%H').date() + \
           datetime.timedelta(days=1)
```

```python
date_air_df = pd.DataFrame()
```

```python
for place, name in interested_places.items():
    place_air_df = air_df[air_df['지역'].str[:2] == name]
    place_air_df['date'] = [my_to_datetime(d) for d in place_air_df['측정일시']]
    place_air_df1 = place_air_df.groupby('date').agg(
        SO2 = pd.NamedAgg(column="SO2", aggfunc='mean'), 
        CO = pd.NamedAgg(column="CO", aggfunc='mean'),
        O3 = pd.NamedAgg(column="O3", aggfunc='mean'),
        NO2 = pd.NamedAgg(column="NO2", aggfunc='mean'),
        PM10 = pd.NamedAgg(column="PM10", aggfunc='mean'),
        PM25 = pd.NamedAgg(column='PM25',aggfunc='mean')
    )
    place_air_df1['place'] = place
    date_air_df = pd.concat([date_air_df, place_air_df1])
```
