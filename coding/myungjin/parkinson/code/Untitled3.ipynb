```python
import pandas as pd

air_1_df = pd.read_csv("../data/2016_1.csv", encoding='EUC-KR')

import datetime
def my_to_datetime(date_str):
    date_str = str(date_str)
    if date_str[8:10] != '24':
        return pd.to_datetime(date_str, format='%Y%m%d%H').date()

    date_str = date_str[0:8] + '00' + date_str[10:]
    return pd.to_datetime(date_str, format='%Y%m%d%H').date() + \
           datetime.timedelta(days=1)

seoul_air_df = air_1_df[air_1_df['지역'] == '서울']

seoul_air_df['date'] = [my_to_datetime(d) for d in seoul_air_df['측정일시']]

seoul_air_df.head(2)

seoul_air_df1 = seoul_air_df.groupby('date').agg(
    place = pd.NamedAgg(column='지역', aggfunc='first'),
    SO2 = pd.NameAgg(column="SO2", aggfunc='average')
#     pd_in = pd.NamedAgg(column='sick_cha3', aggfunc=lambda x: x[lambda a: a.isin(['G20','G21','G22'])].count())
#         pd_in = pd.NamedAgg(column='sick_cha3', aggfunc='count')
)
```
